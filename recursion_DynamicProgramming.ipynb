## longest common subsequence

subsequence -> sequence obtained by deleting zero or more elements from another sequence  
Sequence -> group of items with a deterministic ordering. Lists, tuples and ranges are some common sequence types in Python.  

```python
def longest_common_subsequence(seq1, seq2, idx1 = 0, idx2 = 0):
    if idx1 == len(seq1) or idx2 == len(seq2):
        return 0

    elif seq1[idx1] == seq2[idx2]:
        return 1 + longest_common_subsequence(seq1, seq2, idx1 + 1, idx2 + 1)

    else:
        res1 = longest_common_subsequence(seq1, seq2, idx1 + 1, idx2)
        res2 = longest_common_subsequence(seq1, seq2, idx1, idx2 + 1)
        return max(res1, res2)
```

**MEMOIZATION**

since each node (if each index access is represented as a tree) will be accessed twice in worst case scenario (letter mismatch - 2 child nodes) --> we can store it in a dictionary called Memo

```python
def lcs_memoization(seq1, seq2):
    memo = {}
    def recursion(idx1 = 0, idx2 = 0):
        key = (idx1, idx2)
        if key in memo:
            return memo[key]

        elif idx1 == len(seq1) or idx2 == len(seq2):
            memo[key] = 0

        elif seq1[idx1] == seq2[idx2]:
            memo[key] =  1 + recursion(idx1 + 1, idx2 + 1)

        else:
            res1 = recursion(idx1 + 1, idx2)
            res2 = recursion(idx1, idx2 + 1)
            memo[key] = max(res1, res2)

        print(memo)
        return memo[key]

    return recursion(0, 0)
    
```

#### Dynamic Programming

Create a table of size (n1+1) * (n2+1) initialized with 0s, where n1 and n2 are the lengths of the sequences.   
table[i][j] represents the longest common subsequence of seq1[:i] and seq2[:j].

- If seq1[i] and seq2[j] are equal, then `table[i+1][j+1] = 1 + table[i][j]`
- If seq1[i] and seq2[j] are equal, then `table[i+1][j+1] = max(table[i][j+1], table[i+1][j]`  

Time Complexity =  O(N1∗N2).

```python
def lcs_dp(seq1, seq2):
    table = [[0 for i in range(len(seq2) + 1)] for j in range(len(seq1) + 1)]

    for i in range(len(seq1)):
        for j in range(len(seq2)):
            if seq1[i] == seq2[j]:
                table[i + 1][j + 1] = table[i][j] + 1
            elif seq1[i] != seq2[j]:
                table[i + 1][j + 1] = max(table[i + 1][j], table[i][j + 1])

    return table[-1][-1]
```

## 0-1 Knapsack Problem

```python
def knapsack(capacity, weights, values, idx = 0):
    if idx == len(weights):
        return 0

    elif weights[idx] > capacity:
        return knapsack(capacity, weights, values, idx+1)

    else:
        res1 = knapsack(capacity, weights, values, idx+1)  #if item not added
        res2 = values[idx] + knapsack(capacity - weights[idx], weights, values, idx+1) #if item added
        return max(res1, res2)
```

#### Dynamic Programming

```python
def knapsack_dp(capacity, weights, values):
    table = [[0 for _ in range(capacity + 1)] for i in range(len(weights) + 1)]

    for i in range(len(weights)):
        for c in range(1, capacity + 1):
            if weights[i] > c:
                table[i + 1][c] = table[i][c]
            elif weights[i] <= c:
                table[i + 1][c] = max(table[i][c], values[i] + table[i][c - weights[i]])

    return table[-1][-1]
```
