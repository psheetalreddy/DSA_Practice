## Recursion
 - A function that calls itself  
 - conditioned on some parameter -> stops when parameter arrives to base condition  
 - 2 parts --> *Base Case* & *Recursive Call*  
 - Pros
     - Bridges the gap between elegance and complexity
     - Reduces the need for complex loops and auxiliary data structures (Since uses inbuilt stack for storing recursive calls)
     - Can reduce time complexity easily with *Memoization*
     - Works really well with recursive structures like tress and graphs  
 - Cons  
     - Slowness due to CPU overhead  
     - Can lead to out of memory errors / stack overflow exceptions
     - Can be unnecessarily complex if poorly constructed
       
**Call Stack :** Each call the recursion function is stored in a stack called CALL STACK and is then executed from last to first call


```python
#Q1 String Reversal

def stringReversal(string):
    string = string.lower()
    if string == "":
        return ""

    return stringReversal(string[1:]) + string[0]
```

```python
print(stringReversal("sheetal"))
```

```python
#Q2 check if palindrome
def isPalindrome(string):
    if (string == "") or (len(string) == 1):
        return True

    if string[0] == string[-1]:
        return isPalindrome(string[1:(len(string)-1)])

    else:
        return False
```

```python
print(isPalindrome("racecar"))
print(isPalindrome("sheetal"))
```

```python
#Q3 Convert Decimal to Binary
def decimalToBinary(num):
    if num == 0:
        return ""

    return decimalToBinary(num // 2) + str(num % 2)  #remainder obtained when dividing number by 2 is one of the digit in binary notation
```

```python
print(decimalToBinary(233))
```
