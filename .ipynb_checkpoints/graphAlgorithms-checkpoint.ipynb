# Graph Traversals

breadth first -> explores nodes in all directions evenly  
depth first -> explores nodes in single direction at a time

breadth first -> queue FIFO    
depth first -> stack LIFO

n - no. of nodes  
e - no. of edges  
time complexity = O(e)  
space complexity = O(n)  
- normally -> n nodes => n^2 edges
                      => time complexity = O(n^2)

```python
def depth_first_search(graph, source):
    stack = [source]
    while(len(stack) > 0):
        current = stack.pop()
        print(current)
        for neighbor in graph[current]:
            stack.append(neighbor)
            
```

```python
graph = {
    "a" : ["b", "c"],
    "b" : ["d"],
    "c" : ["e"],
    "d" : ["f"],
    "e" : [],
    "f" : []
}

print(depth_first_search(graph, "a"))
```

```python
#using recursion
def dfs_recursive(graph, source):
    print(source)
    for neighbor in graph[source]:
        dfs_recursive(graph, neighbor)
```

```python
print(dfs_recursive(graph, "a"))
```

```python
#STACK
l = []
for i in range(5):
    l.append(i)
    print(l)
for i in range(5):
    l.pop()
    print(l)
```

```python
#QUEUE
from collections import deque

# enqueue
q = deque()
for i in range(5):
    q.append(i + 2)
    print(list(q))

# dequeue
for i in range(5):
    q.popleft()
    print(list(q))
```

**Find if Path exists between source and destination**  
can use both dfs and bfs

```python
#DEPTH FIRST SEARCH
def has_path(graph, src, dest):
    if src == dest:
        return True

    for neighbor in graph[src]:
        if (has_path(graph, neighbor, dest)):
            return True
    return False
```

```python
print(has_path(graph, "a", "d"))
```

```python
#BREADTH FIRST SEARCH
from collections import deque

def has_path(graph, src, dest):
    queue = deque()
    queue.append(src)
    while len(queue) > 0:
        current = queue.popleft()
        for neighbor in graph[current]:
            queue.append(neighbor)
            if neighbor == dest:
                return True
```

```python
graph1 = {
    "a" : ["c", "b"],
    "b" : ["d"],
    "c" : ["e"],
    "d" : ["f"],
    "e" : [],
    "f" : []
}

print(has_path(graph1, "a", "d"))
```

### Undirected Graphs

given edges(lis of lists) convert to dictionary(node - connected nodes) -> both nodes(as keys) should have the other node as value

```python
#check if path exists when undirected graph is given
def undirectedGraph(edges, nodeA, nodeB):
    graph = buildGraph(edges) #convert undirected edges to graph representation
    #print(graph)
    res = hasPathDFS(graph, nodeA, nodeB)

    if res:
        print("Path Exists")
    else:
        print("Path Does not exist")
    

def buildGraph(edges):
    graph = {}

    for edge in edges:
        [a, b] = edge
        if(a not in graph.keys()):
            graph[a] = []
        if(b not in graph.keys()):
            graph[b] = []
        graph[a].append(b)
        graph[b].append(a)
        
    return graph

def hasPathDFS(graph, src, dest):
    try:
        if src == dest:
            return True
        for neighbor in graph[src]:
            if hasPathDFS(graph, neighbor, dest):
                return True
    except RecursionError:
        print("does not exist")
    finally:
        return False
```

```python
edges1 = [
    ['i', 'j'],
    ['k', 'i'],
    ['m', 'k'],
    ['k', 'l'],
    ['o', 'n']
]
```

```python
undirectedGraph(edges1, 'i', 'j')
```
