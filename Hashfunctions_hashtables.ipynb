**Problem Statement - Python Dictionaries and Hash Tables**  
Recreate Python dictionaries from scratch using data structure -> hash table.  
Dictionaries in Python are used to store key-value pairs. Keys are used to store and retrieve values.  
Dictionary in python  -> uses hash tables

```python
#insert logic -> take sum of ord(each character) % 4096(size of hash table) = key of that value
def get_index(data_list, a_string):
    result = 0
    for ch in a_string:
        result += ord(ch)

    #here we aren't using MAX_HASH_TABLE_SIZE(a global variable with standard hash table size = 4096) but using length of data
    #this function can now be applied on data of any size
    #function which only uses local variables and arguments (no external/global/fixed variables) --> PURE FUNCTION
    index = result % len(data_list)
    return index
```

**Basic Hash Table**

```python
class BasicHashTable:
    def __init__(self, size = 4096): #default argument is 4096 but we can give any value if needed
        self.size = size
        self.data_list = [None] * self.size

    def get_index(self, key):
        result = 0
        for i in key:
            result += ord((i))
        index = result % self.size

        if (self.data_list[index] is not None and self.data_list[index][0] == key):
            return index

        while(self.data_list[index]):
            index = (index + 1)
            index = index  % len(self.size)
        
        return index

    def insert(self, data):
        k, v = data
        index = self.get_index(k)
        self.data_list[index] = data
        print(f'{data} stored in {index}')

    def find(self, key):
        index = self.get_index(key)
        pair = self.data_list[index]
        if pair == None:
            return "Does not exist"
        else:
            return pair

    def update(self, key, value):
        index = self.get_index(key)
        new_kv = key, value
        self.data_list[index] = new_kv

    def list(self):
        datas = [kv for kv in self.data_list if kv is not None]
        for i in datas:
            print(f"key = {i[0]} value = {i[1]}")
```

```python
hash_table = BasicHashTable()
```
