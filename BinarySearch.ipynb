```python
#Q1 find the starting index and ending index of a given number(repeating itself in a sorted array(ascending))

def binarySearch(array, number, find_first):
    lo = 0
    hi = len(array) - 1
    result = -1  

    while(lo <= hi):
        mid_pos = (lo + hi) // 2
        mid_val = array[mid_pos]

        if(mid_val == number):
            result = mid_pos
            if(find_first == True):
                hi = mid_pos - 1  
            else:
                lo = mid_pos + 1  

        elif(mid_val > number):
            hi = mid_pos - 1
        else:
            lo = mid_pos + 1

    return result


def firstPosition(array, number):
    return binarySearch(array, number, find_first = True)

def lastPosition(array, number):
    return binarySearch(array, number, find_first = False)

def findIndex(array, number):
    return (firstPosition(array, number), lastPosition(array, number))
```
